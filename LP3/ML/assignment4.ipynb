```python
x = 2
lr = 0.01
precision = 0.000001
previos_step_size = 1
max_iter = 10000
iters = 0
gf = lambda x : x**2 + 18*x + 9
gd = []
```

```python
while precision < previos_step_size and iters < max_iter:
    prev = x
    x  = x - lr*gf(prev)
    previos_step_size = abs(x - prev)
    iters +=1
    gd.append(x)
    print("Iterations" , iters, 'value' , x )
```

```python
from matplotlib import pyplot as plt
plt.plot(range(70), gd)
```
